```python
import torch
import torch.nn as nn
import torch.nn.functional as f
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
X,y= load_breast_cancer(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
y_train
```

Converting numpy data to tensor data

```python
X_train_scaled_tensor=torch.from_numpy(X_train_scaled).float()
X_test_scaled_tensor=torch.from_numpy(X_test_scaled).float()
y_train_tensor=torch.from_numpy(y_train).float().unsqueeze(1)
y_test_tensor=torch.from_numpy(y_test).float().unsqueeze(1)
```

creating train dataset

```python
train_dataset=TensorDataset(X_train_scaled_tensor,y_train_tensor)
```

creating loader dataset

```python
X_test_scaled_tensor.shape
```

```python
train_loader= DataLoader(train_dataset, batch_size=15, shuffle=True )
```

Creating or defining neural network

```python
class BCNet(nn.Module):
  def __init__(self):
    super(BCNet,self).__init__()
    self.fc1=nn.Linear(30,40)
    self.fc2=nn.Linear(40,30)
    self.fc3=nn.Linear(30,1)

  def forward(self,x):
    x=f.relu(self.fc1(x))
    x=f.relu(self.fc2(x))
    x=f.sigmoid(self.fc3(x))

    return x
```

```python
model=BCNet()
```

loss function and optimizer

```python
criterion=nn.BCELoss()
optimizer=optim.Adam(model.parameters(),lr=0.001)
```

Training

```python
epochs=20

for epoch in range(epochs):
  model.train()
  running_loss=0.0

  for x_batch,y_batch in train_loader:
    optimizer.zero_grad()
    preds=model(x_batch)
    loss=criterion(preds,y_batch)

    loss.backward()
    optimizer.step()

    running_loss += loss.item()

  print(f'Epoch {epoch+1}: loss was {(running_loss)/len(train_loader)}')

```

Evaluation

```python
with torch.no_grad():
  model.eval()

  preds=model(X_test_scaled_tensor)
  loss=criterion(preds, y_test_tensor).item()

  accuracy=((preds>=0.5)== y_test_tensor).float().mean().item()
```
